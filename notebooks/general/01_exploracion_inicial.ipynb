```python
# Ya en materia para lo que nos interesa. Shotchartdetail nos muestra las estadísticas de los lanzamientos

# Primero sacamos los datos de 2024-25 que es como aparecen por temporadas.
df_shots = pd.DataFrame(shotchartdetail.ShotChartDetail(team_id=0, player_id=0, season_nullable="2024-25").get_data_frames()[0])

df_shots
```

```python
# Pero solo nos interesan estas dos columnas, que son las coordenadas desde dónde se hicieron los lanzamientos
df_shots_coords = pd.DataFrame(df_shots[["LOC_X", "LOC_Y"]])
df_shots_coords
```

```python
# Aquí vemos que los lanzamientos "tienen pinta" de tener una tendencia.
plt.figure(figsize=(8,10))

sns.scatterplot(data=df_shots_coords,
                x="LOC_X", y="LOC_Y",)

plt.show()
```

```python
# Si hacemos mucho más grande el Figure, se ve mejor la tendencia.
# Se dibuja lo que esperábamos

plt.figure(figsize=(35,50))

sns.scatterplot(data=df_shots_coords,
                x="LOC_X", y="LOC_Y",)

plt.show()
```

```python
# Guardamos SOLAMENTE los lanzamientos del 2025 filtrando a través del GAME_DATE
# Si leemos los datos, vemos que todos empiezan por el año. Facilísimo de extraer.

df_shots["GAME_DATE"]

df_shots_2025 = df_shots[df_shots["GAME_DATE"].astype(str).str.startswith("2025")]

df_shots_2025
```

```python
# Gráfico para demostrar que se cumple el inicio de nuestra hipótesis. En 2025 la tendencia es la que se ve en el gráfico.
# Con el mapa de calor se ve muy claro.

# Importamos mcolors para que se vea bonito
import matplotlib.colors as mcolors

fig, ax = plt.subplots(figsize=(50,50))

x = df_shots_2025["LOC_X"]
y = df_shots_2025["LOC_Y"]

hb = ax.hexbin(
    x, y,
    gridsize=50,
    cmap="hot",
    mincnt=0,
    norm=mcolors.PowerNorm(gamma = 0.45),
    extent=[-250, 250, -50, 450]
)

plt.title("Mapa de calor de lanzamientos - 2025", fontsize=112)
plt.axis("off")
plt.tight_layout()

plt.show()
```
